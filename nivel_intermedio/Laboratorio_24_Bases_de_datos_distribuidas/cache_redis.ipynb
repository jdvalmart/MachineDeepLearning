```python
import redis
import pandas as pd
```

```python
# Configurar conexión a Redis
client = redis.StrictRedis(host='localhost', port=6379, decode_responses=True)
```

```python
# Cargar el dataset Movielens
df = pd.read_csv('movies.csv') # Asegúrese de descargar previamente el archivo
```

```python
# Función para cargar datos en Redis
def cache_data():
    for _, row in df.iterrows():
        # Almacenar datos como clave-valor
        client.set(row['movieId'], row['title'], ex=30)  # Expira en 30 segundos

# Recuperar un título específico desde Redis
def get_movie_title(movie_id):
    title = client.get(movie_id)
    if title:
        print(f"Película encontrada en cache: {title}")
    else:
        print("Película no encontrada en cache.")

# Ejecutar
cache_data()
get_movie_title('1') # Cambie el ID para probar diferentes película
```
