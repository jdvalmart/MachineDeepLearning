## Ejercicio 1: Ventas mensuales

Creamos un gráfico de barras con las ventas de cada mes y trazamos una línea horizontal que representa la **media**. Esto permite identificar visualmente qué meses están por encima o por debajo del promedio.

**Conceptos aplicados:**
- `pd.DataFrame`: estructura tabular de pandas
- `mean()` y `std()`: estadísticos descriptivos
- `plt.bar()`: gráfico de barras
- `plt.axhline()`: línea de referencia horizontal

```python
data = {
    "Mes":["Enero","Febrero","Marzo","Abril","Mayo","Junio","Julio","Agosto","Septiembre","Octubre","Noviembre","Diciembre"],
    "Ventas":[25,30,28,45,50,63,56,90,36,73,85,12]
    }

df = pd.DataFrame(data)
media = df["Ventas"].mean()
desviacion = df["Ventas"].std()

df["es_mayor_a_la_media"]= df['Ventas'] > media
plt.bar(df["Mes"],df["Ventas"], color="orange")
plt.title("Ventas Mensuales")
plt.xlabel("Meses")
plt.ylabel("Ventas")
plt.xticks(rotation=90)
plt.axhline(media, color="red", linestyle="--", label="Media de ventas")
```

## Ejercicio 2: Promedio de notas

Calculamos el promedio de tres materias para cada estudiante y graficamos los resultados. Además identificamos quiénes tienen promedio superior a 90.

**Conceptos aplicados:**
- Operaciones entre columnas de un DataFrame
- `mean(axis=1)`: promedio por fila (por estudiante)
- Filtros booleanos: `df["Promedio"] > 90`

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


new_data = {
    "Estudiante":["Juan","Camila","Lenin","Mariana", "Yolanda", "Carolina", "Yadira", "Alirio", "Emmanuel", "Tomate"],
    "Matematicas":[85,98,70,60,45,95,96,75,18,59],
    "Ciencias":[96,85,75,67,82,49,50,68,81,78],
    "Historia":[78,98,34,67,86,90,78,54,67,83]
}

df2 = pd.DataFrame(new_data)

media_matematicas = df2["Matematicas"].mean()
desv_matematicas = df2["Matematicas"].std()

media_ciencias = df2["Ciencias"].mean()
desv_ciencias = df2["Ciencias"].std()

media_historia = df2["Historia"].mean()
desv_historia = df2["Historia"].std()

df2["Promedio"] = df2[["Matematicas","Ciencias","Historia"]].mean(axis=1)
df2["Promedio_mayor_a_90"] = df2["Promedio"] > 90

plt.bar(df2["Estudiante"], df2["Promedio"], color="red")
plt.title("Promedio de notas del estudiante")
plt.xlabel("Estudiante")
plt.ylabel("Promedio")
plt.xticks(rotation=90)
plt.show()

```
