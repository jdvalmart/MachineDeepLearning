# Laboratorio 10 — Scikit-learn y Seaborn
## Objetivo
Aprender a usar **scikit-learn** para entrenar un modelo de Machine Learning (Regresión Logística) y **seaborn** para visualizar los resultados sobre el clásico dataset **Iris**.

Este es tu primer pipeline completo de ML: cargar datos → entrenar modelo → predecir → evaluar → visualizar.

## 1. Importación de librerías

Usamos:
- `sklearn.datasets` para cargar el dataset Iris
- `sklearn.model_selection` para dividir en entrenamiento y prueba
- `sklearn.linear_model` para el modelo de Regresión Logística
- `sklearn.metrics` para medir la exactitud
- `seaborn` para visualización avanzada

```python
# Importar las librerias
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# invocar las librerias
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Cargar los datos
data = load_iris()

X = data.data[:, :2]  # Usamos solo las dos primeras características para visualización
y = data.target

# División en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Entrenamiento del modelo
model = LogisticRegression(max_iter=200)  # Se recomienda aumentar iteraciones
model.fit(X_train, y_train)

# Predicción
y_pred = model.predict(X_test)

# Evaluación
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")

# Crear DataFrame para graficar
df = pd.DataFrame({
    "Longitud del sepalo": X_test[:, 0],
    "Ancho del sepalo": X_test[:, 1],
    "EspecieReal": y_test,
    "EspeciePredicha": y_pred
})

# Graficar las predicciones
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x="Longitud del sepalo",
    y="Ancho del sepalo",
    hue="EspeciePredicha",
    data=df,
    palette="Dark2",
    marker="X",
    s=100
)

plt.title("Predicciones de especies de flores")
plt.xlabel("Longitud del sépalo")
plt.ylabel("Ancho del sépalo")
plt.legend(title="Especie predicha")
plt.show()

print("Predicciones de especies de flores:")
print(df.head())

```

## 2. Pipeline completo de ML

El flujo de trabajo estándar en Machine Learning:

1. **Cargar datos** → `load_iris()`
2. **Separar características (X) y etiquetas (y)**
3. **Dividir en train/test** → `train_test_split()`
4. **Entrenar modelo** → `LogisticRegression().fit()`
5. **Predecir** → `model.predict()`
6. **Evaluar** → `accuracy_score()`
7. **Visualizar** → `sns.scatterplot()`

**Nota:** Usamos solo 2 características (largo y ancho del sépalo) para poder visualizar en 2D.

## Visualización con Seaborn

El `scatterplot` de seaborn colorea cada punto según la especie predicha por el modelo. Esto permite ver de un vistazo cómo el modelo separa las 3 clases de flores Iris (setosa, versicolor, virginica).

**¿Por qué seaborn y no matplotlib directamente?** Seaborn ofrece una API de más alto nivel: maneja automáticamente la paleta de colores, la leyenda y el estilo visual.
