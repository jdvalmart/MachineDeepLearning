# Análisis de sentimiento


**Actividad 1**

```python
from textblob import TextBlob
```

```python
frases = [
    "The weather today is absolutely beautiful!",
    "I am really disappointed with the new update.",
    "The event was well-organized but lacked excitement."
]

etiquetas_reales = [1, -1, 0]

predicciones = []

for frase in frases:
    blob = TextBlob(frase)

    print(f"Frase: '{frase}' | Sentimiento: {blob.sentiment.polarity}" )

    if blob.sentiment.polarity > 0:
       predicciones.append(1)
    elif blob.sentiment.polarity < 0:
       predicciones.append(-1)
    else:
       predicciones.append(0)
```

```python
from sklearn.metrics import accuracy_score
precision = accuracy_score(etiquetas_reales, predicciones)
print(f"Precision de analisis de sentimiento con textblob: {precision}")
```

**Actividad 2**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split

from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

import matplotlib.pyplot as plt
import seaborn as sns


frases = [
    "This app is very intuitive!",
    "I didn't enjoy the customer service.",
    "Quite average experience.",
    "Fantastic presentation!",
    "The food was awful."
]

sentimientos = [1, 0, 0, 1, 0]
```

```python
vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(frases)
X_train, X_test, y_train, y_test = train_test_split(X, sentimientos, test_size=0.2, random_state=42)

modelo = MultinomialNB()

modelo.fit(X_train, y_train)
```

```python
y_pred = modelo.predict(X_test)

precision = accuracy_score(y_test, y_pred)
print(f"Precision del analisis de sentimiento con Naive Bayes: {precision}")
```

```python
reporte_clasificacion = classification_report(y_test, y_pred)
print("Reporte de Clasificación:")
print(reporte_clasificacion)
```
